## 🎯 **Project Objectives**

This project aims to perform a comprehensive Exploratory Data Analysis (EDA) on the Amazon Sales Report. The primary objectives are to:

*   **Understand Sales Performance:** Analyze overall sales trends, revenue distribution, and quantity sold across various dimensions.
*   **Identify Key Influencers:** Determine the most impactful factors on sales, such as product category, sales channel, fulfillment method, and geographical location.

## 📊 **Data Explanation**

The dataset used for this analysis is the **Amazon Sales Report**. It contains detailed transactional information regarding sales activities on the Amazon platform. Key columns include:

*   **Order ID:** Unique identifier for each order.
*   **Date:** Date of the transaction.
*   **Status:** Current status of the order (e.g., Shipped, Cancelled).
*   **Fulfilment:** Method of fulfillment (e.g., Amazon, Merchant).
*   **Sales Channel:** The channel through which the sale occurred (e.g., Amazon.in).
*   **Category:** Product category.
*   **Size:** Product size.
*   **Qty:** Quantity of items sold.
*   **Amount:** Revenue generated from the sale.
*   **ship-city, ship-state, ship-postal-code, ship-country:** Geographical details of the shipping destination.
*   **B2B:** Indicates if the order is Business-to-Business.

This dataset provides a rich source of information to understand customer purchasing behavior, product popularity, and logistical performance within the Amazon ecosystem.

### Geocoding Top 20 Ship Cities

To create the map visualization, we need geographical coordinates for each city. I will now geocode only the top 20 revenue-generating cities to manage the number of requests to the geocoding service. This will convert the city names into latitude and longitude coordinates.

```python
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
import plotly.express as px
import time

geolocator = Nominatim(user_agent="geoapi_colab_project")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)

# Get top 20 unique cities to geocode
top_20_cities_revenue = ship_city_revenue.head(20)
unique_cities_top_20 = top_20_cities_revenue['ship-city'].unique()

# Create a dictionary to store geocoded results
city_coords_top_20 = {}

print(f"Geocoding {len(unique_cities_top_20)} unique cities. This might take some time...")

for city in unique_cities_top_20:
    try:
        location = geocode(city + ', India') # Specifying India for better accuracy
        if location:
            city_coords_top_20[city] = {'latitude': location.latitude, 'longitude': location.longitude}
        else:
            city_coords_top_20[city] = {'latitude': None, 'longitude': None}
    except Exception as e:
        print(f"Error geocoding {city}: {e}")
        city_coords_top_20[city] = {'latitude': None, 'longitude': None}

# Convert to DataFrame for merging
geocoded_df_top_20 = pd.DataFrame.from_dict(city_coords_top_20, orient='index')
geocoded_df_top_20 = geocoded_df_top_20.reset_index().rename(columns={'index': 'ship-city'})

# Merge with top_20_cities_revenue
ship_city_revenue_geo = pd.merge(top_20_cities_revenue, geocoded_df_top_20, on='ship-city', how='left')

# Drop rows where geocoding failed (latitude/longitude is None)
ship_city_revenue_geo.dropna(subset=['latitude', 'longitude'], inplace=True)

display(ship_city_revenue_geo.head())
```

### Ship City Wise Total Amount Revenue Map Visual (Top 20 Cities)

Here is the interactive map visualization of the total revenue for the top 20 ship cities. The size and color of the circles on the map represent the total revenue, allowing for a quick visual understanding of revenue distribution across these major cities.

```python
fig = px.scatter_mapbox(
    ship_city_revenue_geo,
    lat="latitude",
    lon="longitude",
    size="Amount",
    color="Amount", # Color by revenue for visual impact
    hover_name="ship-city",
    hover_data={
        "Amount": ':.2f',
        "latitude": False,
        "longitude": False
    },
    zoom=4, # Adjust zoom level to focus on India and top cities
    height=600,
    title='Ship City Wise Total Amount Revenue (Top 20 Cities)',
    mapbox_style="carto-positron" # A light map style
)

fig.update_layout(margin={"r":0,"t":50,"l":0,"b":0})
fig.show()
```
